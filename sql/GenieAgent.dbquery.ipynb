```sql
-- Los 10 Sismos de Mayor Magnitud Registrados
SELECT
    `FECHA_UTC`,
    `MAGNITUD`,
    `LATITUD`,
    `LONGITUD`,
    `PROFUNDIDAD`
FROM `big_data_team1`.`saori`.`mis_datos_igp`
WHERE `MAGNITUD` IS NOT NULL
ORDER BY `MAGNITUD` DESC
LIMIT 10

-- Estadísticas Generales de Sismos
SELECT
    COUNT(*) AS `total_sismos`,
    ROUND(AVG(`MAGNITUD`), 2) AS `magnitud_promedio`,
    MIN(`MAGNITUD`) AS `magnitud_minima`,
    MAX(`MAGNITUD`) AS `magnitud_maxima`
FROM `big_data_team1`.`saori`.`mis_datos_igp`
WHERE `MAGNITUD` IS NOT NULL


-- Estadísticas Completas por Año
SELECT
    YEAR(`FECHA_UTC`) AS `año`,
    COUNT(*) AS `cantidad_sismos`
FROM `big_data_team1`.`saori`.`mis_datos_igp`
WHERE `FECHA_UTC` IS NOT NULL
GROUP BY YEAR(`FECHA_UTC`)
ORDER BY `cantidad_sismos` DESC


-- Distribución de Sismos por Rango de Magnitud
SELECT
    CASE
        WHEN `MAGNITUD` < 4 THEN 'Menor a 4'
        WHEN `MAGNITUD` >= 4 AND `MAGNITUD` < 5 THEN 'Entre 4 y 5'
        WHEN `MAGNITUD` >= 5 AND `MAGNITUD` < 6 THEN 'Entre 5 y 6'
        WHEN `MAGNITUD` >= 6 THEN 'Mayor o igual a 6'
    END AS `rango_magnitud`,
    COUNT(*) AS `cantidad_sismos`,
    ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER(), 2) AS `porcentaje`
FROM `big_data_team1`.`saori`.`mis_datos_igp`
WHERE `MAGNITUD` IS NOT NULL
GROUP BY
    CASE
        WHEN `MAGNITUD` < 4 THEN 'Menor a 4'
        WHEN `MAGNITUD` >= 4 AND `MAGNITUD` < 5 THEN 'Entre 4 y 5'
        WHEN `MAGNITUD` >= 5 AND `MAGNITUD` < 6 THEN 'Entre 5 y 6'
        WHEN `MAGNITUD` >= 6 THEN 'Mayor o igual a 6'
    END
ORDER BY
    CASE
        WHEN `rango_magnitud` = 'Menor a 4' THEN 1
        WHEN `rango_magnitud` = 'Entre 4 y 5' THEN 2
        WHEN `rango_magnitud` = 'Entre 5 y 6' THEN 3
        WHEN `rango_magnitud` = 'Mayor o igual a 6' THEN 4
    END


-- Profundidad Promedio de Sismos por Rango de Latitud
SELECT
    ROUND(`LATITUD`, 0) AS `latitud_redondeada`,
    ROUND(AVG(`PROFUNDIDAD`), 2) AS `profundidad_promedio`,
    COUNT(*) AS `cantidad_sismos`
FROM `big_data_team1`.`saori`.`mis_datos_igp`
WHERE `LATITUD` IS NOT NULL
    AND `PROFUNDIDAD` IS NOT NULL
GROUP BY ROUND(`LATITUD`, 0)
ORDER BY `latitud_redondeada` DESC


-- Validación de Calidad de Datos Sísmicos
SELECT
    COUNT(*) AS `total_sismos`,
    SUM(CASE WHEN `DENTRO_RANGO_PERU` = TRUE THEN 1 ELSE 0 END) AS `dentro_rango_peru_true`,
    ROUND(SUM(CASE WHEN `DENTRO_RANGO_PERU` = TRUE THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS `porcentaje_dentro_peru`,
    SUM(CASE WHEN `PROFUNDIDAD_VALIDA` = FALSE THEN 1 ELSE 0 END) AS `profundidad_valida_false`,
    ROUND(SUM(CASE WHEN `PROFUNDIDAD_VALIDA` = FALSE THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS `porcentaje_profundidad_invalida`
FROM `big_data_team1`.`saori`.`mis_datos_igp`
```
