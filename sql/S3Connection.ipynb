```python
import re

# Solo cambias esta línea con cada archivo nuevo
ruta = "s3://igp-team01/integracion_prueba3.csv"

# El nombre de la tabla sale del nombre del archivo
nombre_archivo = ruta.split("/")[-1].replace(".csv", "")
nombre_limpio = re.sub(r"[^a-zA-Z0-9_]", "_", nombre_archivo).lower()
nombre_tabla = f"big_data_team1.saori.{nombre_limpio}"

df = (
    spark.read
    .option("header", "true")
    .option("delimiter", ";")
    .option("inferSchema", "true")
    .option("encoding", "UTF-8")
    .csv(ruta)
)

# Por si las columnas traen espacios o caracteres raros, que Delta no los acepta
for c in df.columns:
    df = df.withColumnRenamed(c, re.sub(r"[^a-zA-Z0-9_]", "_", c))

df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(nombre_tabla)

print(f"¡Listo! Tabla guardada con éxito en: {nombre_tabla}")
```
