```python
# 1. Ruta de tu archivo en el S3
ruta = "s3://igp-team01/igp_2026-10-20 (1).csv"

# 2. Lectura del CSV indicando que el separador es ';'
df = (
    spark.read
    .option("header", "true")
    .option("delimiter", ";")  # <--- Esto separa correctamente las columnas
    .option("inferSchema", "true")
    .option("encoding", "UTF-8")
    .csv(ruta)
)

# 3. Guardar en el catálogo 'big_data_team1' y esquema 'saori'
nombre_tabla = "big_data_team1.saori.mis_datos_igp"

df.write.mode("overwrite").saveAsTable(nombre_tabla)

print(f"¡Listo! Tabla guardada con éxito en: {nombre_tabla}")
```
